```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("omarstat2/supply-chain-logistics-data")

print("Path to dataset files:", path)
```

```python
import os

# Ver qué archivos hay
print("\nArchivos descargados:")
for f in os.listdir(path):
    print(f" - {f}")
```
